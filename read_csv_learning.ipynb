#Learning Pandas read_csv()

```python
#Importing pandas
import pandas as pd
```

Reading the CSV File using pandas

Syntax: pd.read_csv(filepath_or_buffer='',sep='',header='',names=[''],dtype={},usecols=[''],skiprows='',index_col='',parse_dates='')

#Parameter Definition:

1. Sep        = Used to sepcify the delimiter, comma(',') it is set as a default value for read_csv function, if you wanted to specify other delimeters you can also specify it here like '|' or '\t'.
2. header     = Used to identify which row to be used as a header (column), Here we specify the index of the rows. index [0] that means first row is considered as the defaul header. 
3. names      = Used to name and rename the column names.
4. dtype      = USed to specify the datatype for the columns manually.
5. usecols    = Used to list down the list of columns instead of getting all columns from the file.
6. skiprows   = Used to skip the unwanted rows from the file.
7. index_col = Used to set the row index for the Dataframe. 0,1,2,3... is initially set as the index for the all the individual rows, If in case if you wanted to set any column as an index you can specify that column name here then that column is treated as the index for the dataframe rows.
8. parse_dates = Use to convert the str object to dataframe object.

Problem-1: Reading an CSV File named Testing.csv


Business Requirement:
1. There is no header or column names specified in the excel, You have to create the column names. Column names=['ID','First_Name','Age']
2. The CSV file containes few unwanted comments in the top of the file, Skip those things and read the data alone.
3. Use the Id field as the Index for all the rows.
4. Manually set the datatype for each columns[Id:Int, First_Name=Str, Age=int]
5. Add the new column named [Email] and set the default value as email='manubarath33@gmail.com' and manually set the datatype=string

```python
#Soltion:

#Step-1: Reading the csv file according to the requirement:
df=pd.read_csv(
            filepath_or_buffer='Testing.csv', 
            sep=',',
            header=None,
            names=['ID', 'First_Name', 'Age'],
            skiprows=3,
            dtype={
                'ID':int,
                'First_Name':'string',
                'Age':int
            },
            index_col=['ID'],
            )

#Step-2: Add the new column named Email".
email_id = 'manubarath33@gmail.com'
df['Email_Id'] = email_id

#Step-3: Manually seting the datatyoe for the newly created column:
df['Email_Id'] = df['Email_Id'].astype('string')

df.head()
```

Proble-2: Reading an CSV file Named Testing_csv.csv

Business Requirement: 
1. There is no header for this csv file, You have to manually create the columns. Column names=['Id','First_Name','Age','DOB'].
2. The CSV file containes few unwanted comments in the top of the file, Skip those things and read the data alone.
3. Manually set the datatype for each columns[Id:Int, First_Name=Str, Age=int]
4. Add an additional new column named [Last_Name] please Note that the column should be created next to the column ['First_Name'] not at the last of the df.
5. Manually set the datatype as Datetime for the column ['DOB']

Note: The most important thing the delimiter for this file is not comma(',') it is pipe('|').

```python
#Solution-2:

#Step-1: Reading the CSV File with respect to th business requirement:
df_csv = pd.read_csv(
        filepath_or_buffer='Testing_Csv.csv',
        sep='|',
        skiprows=2,
        header=None,
        names=['ID','First_Name','Age','DOB'], 
        dtype={
            'ID':int,
            'First_Name':'string',
            'Age':'string'
        } 
        )

#Step-2: Converting the datatype from str to Date
df_csv['DOB'] = pd.to_datetime(df_csv['DOB'], format='%d-%m-%Y', errors='coerce')

#Step-3: Creating a new column named [Last_Name] next to [First_Name]
LastName=['Kanakarajan','Balamani']
df_csv.insert(2,'Last_Name',LastName)

df_csv['Last_Name']=df_csv['Last_Name'].astype('string')

#df_csv.head()

df_csv.dtypes
```
